# 5. Data Quality Analysis

Before performing analysis, investigate missing values, duplicates, data types and unusual values.

```python
# Missing values

missing = df.isnull().sum().sort_values(ascending=False)

missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percentage.round(2)
})

# Keep only columns having missing values
missing_summary = missing_summary[missing_summary["Missing_Count"] > 0]

print("Missing Values Summary:")
print(missing_summary)
```

```python
# Visualize missing values

plt.figure(figsize=(12, 5))

missing_plot = missing[missing > 0]

if len(missing_plot) > 0:
    sns.barplot(x=missing_plot.index, y=missing_plot.values)

    plt.xticks(rotation=60, ha="right")
    plt.ylabel("Number of Missing Values")
    plt.xlabel("Columns")
    plt.title("Missing Values by Column")
    plt.tight_layout()
    plt.show()

else:
    print("No missing values found.")
```

```python
# Duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

# Outlier Investigation

```python
# Outlier Investigation

numeric_columns = df.select_dtypes(include="number").columns

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_summary.append({
        "Column": col,
        "Outlier_Count": outlier_count,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary.sort_values("Outlier_Count", ascending=False))
```

```python
# Visualize outliers using boxplots

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

# Univariate Analysis

```python
#  Univariate Analysis
# Analyze individual variables separately

# 1. Distribution of numerical variables

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
# 2. Distribution of categorical variables

categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    plt.figure(figsize=(10, 5))
    df[col].value_counts().head(10).plot(kind="bar")
    plt.title(f"Top Categories of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

# Bivariate Analysis

```python
# Bivariate Analysis
# Analyze the relationship between two variables

# 1. Crop vs Average Yield

plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Crop", y="Yield_Tonnes_Ha")
plt.title("Yield Distribution by Crop")
plt.xlabel("Crop")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 2. Rainfall vs Yield

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha")
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# 3. Fertilizer Usage vs Yield

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Fertilizer_kg_ha", y="Yield_Tonnes_Ha")
plt.title("Fertilizer Usage vs Crop Yield")
plt.xlabel("Fertilizer (kg/ha)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# 4. Season vs Yield

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Season", y="Yield_Tonnes_Ha")
plt.title("Yield Distribution by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

# Multivariate Analysis

```python
# Multivariate Analysis
# Analyze the relationship between multiple variables

# Compare Yield with Rainfall, Temperature, and Humidity

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season",
    size="Humidity_pct",
    alpha=0.7
)

plt.title("Multivariate Analysis: Rainfall, Yield, Season and Humidity")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Another multivariate analysis:
# Crop, Season and Average Yield

season_crop_yield = (
    df.groupby(["Crop", "Season"])["Yield_Tonnes_Ha"]
      .mean()
      .reset_index()
)

display(
    season_crop_yield
    .sort_values("Yield_Tonnes_Ha", ascending=False)
)
```

```python
# Visualize Crop × Season × Yield

plt.figure(figsize=(12, 6))

sns.barplot(
    data=season_crop_yield,
    x="Crop",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Average Yield by Crop and Season")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# Correlation Analysis

```python
# Correlation Analysis
# Examine correlations between numerical variables

correlation_matrix = df[numeric_columns].corr()

print("Correlation Matrix:")
display(correlation_matrix.round(2))
```

```python
# Visualize the correlation matrix

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# Find the strongest correlations with Yield

yield_correlation = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .drop("Yield_Tonnes_Ha")
    .sort_values(ascending=False)
)

print("Correlation of Numerical Variables with Yield:")
display(yield_correlation.round(2))
```

# Seasonal Comparisons

```python
# Seasonal Comparisons
# Compare important agricultural performance measures across seasons

season_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Rainfall_mm": "mean",
    "Avg_Temperature_C": "mean",
    "Water_Used_m3": "mean",
    "Profit_INR": "mean"
}).round(2)

print("Agricultural Performance by Season:")
display(season_summary)
```

```python
# Visualize average yield by season

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha"
)

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Visualize average profit by season

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Profit_INR"
)

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.tight_layout()
plt.show()
```

# Student-Designed Analysis

```python
#Student-Designed Analysis 1
# Crop-wise performance analysis

crop_summary = df.groupby("Crop").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean"
}).round(2)

print("Crop-wise Performance:")
display(crop_summary.sort_values("Yield_Tonnes_Ha", ascending=False))
```

```python
# Visualize average yield by crop

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Crop",
    y="Yield_Tonnes_Ha",
    order=df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False).index
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#Student-Designed Analysis 2
# State-wise agricultural performance analysis

state_summary = df.groupby("State").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean"
}).round(2)

print("State-wise Agricultural Performance:")
display(state_summary.sort_values("Yield_Tonnes_Ha", ascending=False))
```

```python
# Visualize average yield by state

state_yield = (
    df.groupby("State")["Yield_Tonnes_Ha"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=state_yield.index,
    y=state_yield.values
)

plt.title("Average Crop Yield by State")
plt.xlabel("State")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Student-Designed Analysis 3
# Resource Efficiency Analysis
# Compare water usage with crop yield

resource_efficiency = df.groupby("Crop").agg({
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Yield_Tonnes_Ha": "mean"
}).round(2)

print("Crop-wise Resource Efficiency:")
display(
    resource_efficiency.sort_values(
        "Water_Efficiency_t_per_1000m3",
        ascending=False
    )
)
```

```python
# Visualize water efficiency by crop

efficiency_order = (
    df.groupby("Crop")["Water_Efficiency_t_per_1000m3"]
      .mean()
      .sort_values(ascending=False)
      .index
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Crop",
    y="Water_Efficiency_t_per_1000m3",
    order=efficiency_order
)

plt.title("Average Water Efficiency by Crop")
plt.xlabel("Crop")
plt.ylabel("Water Efficiency (Tonnes per 1000 m³)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
